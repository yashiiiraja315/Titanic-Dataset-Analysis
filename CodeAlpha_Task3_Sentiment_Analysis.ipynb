# **Task 3 – Sentiment Analysis**

**Objective**:

 Analyze text data and classify sentiments into positive, negative, and neutral categories using Natural Language Processing (NLP).

```python
import pandas as pd

data = {
    "review": [
        "This product is amazing and I really loved it.",
        "The service was excellent and very helpful.",
        "I am happy with my purchase.",
        "The product is okay, nothing special.",
        "It was an average experience.",
        "The quality is terrible and disappointing.",
        "I hate this product. It is completely useless.",
        "Very bad service and poor quality."
    ]
}

df = pd.DataFrame(data)

df
```

```python
from textblob import TextBlob

def get_sentiment(text):
    polarity = TextBlob(text).sentiment.polarity

    if polarity > 0:
        return "Positive"
    elif polarity < 0:
        return "Negative"
    else:
        return "Neutral"

df["sentiment"] = df["review"].apply(get_sentiment)

df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x="sentiment", data=df)

plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")
plt.show()
```

# **Key Findings**


• Positive reviews indicate customer satisfaction with the product or service.

• Negative reviews indicate dissatisfaction and areas that may need improvement.

• Neutral reviews represent average or mixed customer experiences.

• Sentiment analysis helps summarize large amounts of text data into understandable categories

# **Conclusion**


Sentiment analysis was used to classify customer reviews into positive, negative, and neutral sentiments. The results provide a simple overview of customer opinions and demonstrate how NLP can be used to analyze text data
